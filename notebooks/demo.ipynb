# Circuit Component Classifier — Demo

Quick walkthrough of the pipeline. All the real logic lives in `src/circuit_classifier/` —
this notebook just calls into it so you can explore results interactively.

```python
import sys, os
sys.path.insert(0, os.path.join('..', 'src'))
sys.path.insert(0, os.path.join('..', 'configs'))

import config as cfg
from circuit_classifier.data import explore_dataset, print_dataset_summary, prepare_data_structure
from circuit_classifier.visualize import visualize_samples, plot_training_history
from circuit_classifier.generators import create_data_generators
from circuit_classifier.model import build_custom_cnn
```

## Explore the dataset

```python
stats, folders = explore_dataset(cfg.RAW_DATASET_PATH)
print_dataset_summary(stats)
```

```python
organized_path, class_names = prepare_data_structure(cfg.RAW_DATASET_PATH, cfg.ORGANIZED_DATA_PATH)
visualize_samples(organized_path, class_names)
```

## Train (short run, for demo purposes)

For a full training run, use `python -m circuit_classifier.train` instead — it's faster
outside the notebook and saves checkpoints/metadata automatically.

```python
from tensorflow.keras.optimizers import Adam

config = dict(cfg.CONFIG)
config['num_epochs'] = 3  # short demo run

train_gen, val_gen = create_data_generators(organized_path, config)

model = build_custom_cnn(input_shape=(*config['img_size'], 1), num_classes=len(class_names))
model.compile(optimizer=Adam(learning_rate=config['learning_rate']),
              loss='categorical_crossentropy', metrics=['accuracy'])

history = model.fit(train_gen, epochs=config['num_epochs'], validation_data=val_gen)
plot_training_history(history)
```
